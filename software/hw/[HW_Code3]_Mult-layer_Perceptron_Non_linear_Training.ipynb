```python
### Training with fancier version ###

import torch
import torch.nn as nn
import torch.nn.functional as F

import torch.optim as optim
import matplotlib.pyplot as plt

class Net(nn.Module): ## nn.Module class is used
    def __init__(self):
        super(Net, self).__init__()
        # 3 layers with bias on
        self.fc1 = nn.Linear(1,64)  # in dim, out dim
        self.fc2 = nn.Linear(64,64)
        self.fc3 = nn.Linear(64,1)
        
    def forward(self, x):
        # realu works but it's very noisy
        ##x = F.relu(self.fc1(x))
        ##x = F.relu(self.fc2(x))
        ##x = F.relu(self.fc3(x))
        ##return x
        # tanh since it's curved it should make things easier
        # although a bit more computational expensive
        x = torch.tanh(self.fc1(x))
        x = torch.tanh(self.fc2(x))
        ##x = torch.tanh(self.fc3(x))
        ##return x
        # wouldn't converge if I did similar to the realu
        return self.fc3(x)

net = Net()

print(net)
print(list(net.parameters())) # parameters are randomized


#def criterion(out, label):
#    return (label - out)**2
criterion = nn.MSELoss()


#optimizer = optim.SGD(net.parameters(), lr=0.1
#to big of steps, and adam should keep momentum
optimizer = optim.Adam(net.parameters(), lr=0.01)


data = [(1.0,3.0), (2.0,6.0), (3.0,10.0), (4.0,15.0), (5.0,22.0), (6.0,32.0)]

# Split data into x and y
x_values, y_values = zip(*data)

# Plotting
plt.figure(figsize=(8, 6))
plt.plot(x_values, y_values, marker='o', linestyle='-', color='b')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.grid(True)
plt.show()
```

```python
for epoch in range(1000): # 0 - 19
    for i, current_data in enumerate(data):
        X, Y = current_data
        X, Y = torch.FloatTensor([X]), torch.FloatTensor([Y])
        optimizer.zero_grad()   
        outputs = net(X)
        loss = criterion(outputs, Y)
        loss.backward()
        optimizer.step()    ## This line is equivalent to "W = W - lr* W.grad"
    if ((epoch % 5) == 0):
        print("Epoch {} - loss: {}".format(epoch, loss))
    # making sure to break as soon as it reaches the min
    if loss.item() < 1e-5: 
        print("Final Epoch {} - loss: {}".format(epoch, loss))
        break


### Test the trained network ###            
for i, current_data in enumerate(data):
    X, Y = current_data
    X, Y = torch.FloatTensor([X]), torch.FloatTensor([Y])  
    out = net(torch.FloatTensor(X))  
    print("when x = {}, y = {}".format(X, out))
    
```
